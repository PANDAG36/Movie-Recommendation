```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("imdb_movies.csv")

a = df.columns

for i in a:
    print(f"{i} = {df[i].isnull().sum()}")

df["genre"] = df["genre"].replace( np.nan, "None")

df["crew"] = df["crew"].replace( np.nan, "None")

print("\n New Datasets \n")

for i in a:
    print(f"{i} = {df[i].isnull().sum()}")
```
